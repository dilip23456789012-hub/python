```python
import pandas as pd
data = {"name":["A","B","C","D"],
        "age":[19,21,18,20],
        "marks":[67,99,80,90],
        }
s=pd.DataFrame(data)
print(s[s["age"] >=20])
print(s[s["marks"] >= 75])
print(s[s["marks"] != 80])
```

```python
import pandas as pd
data = {"name":["A","B","C","D"],
        "age":[19,21,18,20],
        "marks":[67,99,80,90],
        }
s=pd.DataFrame(data)
print(s[(s["marks"] >= 90) &(s["age"] >= 20)] )
```
